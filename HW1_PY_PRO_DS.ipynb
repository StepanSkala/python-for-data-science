```python
import numpy as np
```

```python
# 1. Generate synthetic data:

data = np.random.normal(size=(100, 10), loc = 50, scale = 15)

print(data.shape)
print(data.dtype)
print(data[:5])
```

```python
# 2. Data cleaning:

data[(data < 20) | (data > 80)] = np.nan

print(np.isnan(data).sum())
```

```python
# 3. Handle missing values:

means = np.nanmean(data, axis = 0)
idx = np.where(np.isnan(data))
data[idx] = means[idx[1]]
```

```python
# 4. Data transformation:

# Computing means and stds
means2 = np.mean(data, axis = 0)
stds = np.std(data, axis = 0)

data_standard = (data - means2) / stds
```

```python
# 5. Array indexing and logical operations:

# Computation of quantiles
quant_75 = np.percentile(data_standard, 75, axis = 0)
quant_25 = np.percentile(data_standard, 25, axis = 0)

# Counting values above the 75th percentile
mask_75 = data_standard > quant_75
print(mask_75.sum())

# Replacing the below 25th percentile values
mask_25 = data_standard < quant_25
idx = np.where(mask_25)
data_standard[idx] = quant_25[idx[1]]
```

```python
# 6. Descriptive statistics:

# Computation of the statistics
means_cols = np.mean(data_standard, axis = 0)
medians = np.median(data_standard, axis = 0)
variance = np.var(data_standard, axis = 0)
std_cols = np.std(data_standard, axis = 0)

overall_mean = np.mean(data_standard)
min_cols = np.min(data_standard, axis = 0)
max_cols = np.max(data_standard, axis = 0)

# Prints
print("Column means:", means_cols)
print("Column medians:", medians)
print("Column variances:", variance)
print("Column std devs:", std_cols)
print("Overall mean:", overall_mean)
print("Column minima:", min_cols)
print("Column maxima:", max_cols)
```
